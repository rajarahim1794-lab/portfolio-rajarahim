#Cleaning Data

```python
#filling missing values 
df["director"] = df["director"].fillna("Unknown")
df["cast"] = df["cast"].fillna("Unknown")
df["country"] = df["country"].fillna("Unknown")

```

```python
df.isnull().sum() 
```

```python
df["date_added"] = df["date_added"].fillna("Not Available")
df["rating"] = df["rating"].fillna("Not Rated")
df["duration"] = df["duration"].fillna("Unknown") # filling missing values 
```

```python
df.isnull().sum()
```

**##Find out how many Movies and TV Shows are in the dataset, visualize them, and interpret the results.**

This is classic EDA storytelling. Analysts always ask:

1. What is the distribution of this categorical column?

2. How can I show it visually for easy understanding?

3. What does it tell me about the data?

```python
df["type"].value_counts() #Count each type
```

```python
type_counts = df["type"].value_counts() # we save it in a variable/df
type_counts
```

```python
import matplotlib.pyplot as plt #importing matplotlib library for visualization

type_counts.plot(kind="bar", color=["#1f77b4", "#ff7f0e"]) # chooses bar chart
plt.title("Netflix Content: Movies vs TV Shows") #title for bar chart 
plt.xlabel("Type") # label on x-axis
plt.ylabel("Count") # label on y_axis 
plt.xticks(rotation=0)  # Makes labels horizontal 
plt.show() #Display the chart
```

## Movies vs TV Shows

From the chart above, we can see that:
- Netflix has more Movies (6131) than TV Shows (2673) in this dataset.
- This could indicate a stronger focus on Movies or the dataset being collected over time may have more Movie entries.
- Such insights help understand **content strategy or dataset structure**.


```python
type_percentage = df["type"].value_counts(normalize=True) * 100
type_percentage
```

### Movies vs TV Shows Percentage

Around 70% of the Netflix dataset are Movies, while 30% are TV Shows.
This provides a better understanding of **distribution without looking at raw counts**.


**Find out which countries produce the most content on Netflix.**

```python
country_counts = df["country"].value_counts().head(10)
country_counts
```

```python
import matplotlib.pyplot as plt

country_counts.plot(kind="bar", color="#2ca02c")
plt.title("Top 10 Countries Producing Netflix Content")
plt.xlabel("Country")
plt.ylabel("Number of Shows")
plt.xticks(rotation=45)
plt.show()
```

```python
country_percentage = df["country"].value_counts(normalize=True).head(10) * 100
country_percentage
```

### Top Countries Percentage

The top 10 countries produce a major portion of Netflix content:
- United States dominates with around 35% of shows.
- India, United Kingdom, and Canada follow.
- Helps understand **global content focus**.


**Explore how Netflix content has changed over time.**

```python
year_counts = df["release_year"].value_counts().sort_index()
year_counts
```

```python
year_counts.plot(kind="line", marker="o", color="#d62728")
plt.title("Netflix Content Release Year Trends")
plt.xlabel("Release Year")
plt.ylabel("Number of Shows")
plt.grid(True)
plt.show()
```

### Release Year Trends

- Netflix content gradually increased after 2000.  
- Peaks appear in recent years showing **more content production**.  
- This can indicate **expansion strategy or popularity growth**.


## Conclusion

- The Netflix dataset has more Movies (70%) than TV Shows (30%).  
- The United States is the dominant content producer, followed by India, the UK, and Canada.  
- Content production increased significantly in recent years, indicating Netflix's expansion.  

This analysis provides a clear understanding of Netflix content distribution,
global focus and growth over time.
